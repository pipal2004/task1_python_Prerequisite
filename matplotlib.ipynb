```python
#Create a sample dataset for Line plot
import matplotlib.pyplot as plt
students = ["A", "B", "C", "D", "E", "F", "G"]
marks = [78, 85, 69, 92, 74, 88, 95]
plt.plot(students, marks, marker='o', color='green', linestyle='--', label="Marks Trend")
for i, mark in enumerate(marks):
    plt.annotate(str(mark),
                 (students[i], mark),
                 textcoords="offset points",
                 xytext=(0, 10),
                 ha='center',
                 fontsize=10,
                 color='black')

plt.title("Line Plot: Student Marks", fontsize=14, fontweight='bold')
plt.xlabel("Students")
plt.ylabel("Marks")
plt.grid(True, linestyle="--", alpha=0.6)
plt.legend()
plt.show()
```

```python
#Create a sample dataset for Bar plot
bars = plt.bar(students, marks, color='orange', edgecolor="black", label="Marks")
for bar in bars:
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 1,
             str(bar.get_height()), ha='center', va='bottom', fontsize=10, fontweight="bold")
plt.title("Bar Plot: Student Marks")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.grid(axis='y', linestyle="--", alpha=0.6)
plt.legend()
plt.show()
```

```python
#Create a sample dataset for Scatter  plot
plt.scatter(students, marks, color='b', marker='*', s=100, label="Marks")
for i, mark in enumerate(marks):
    plt.annotate(str(mark),
                 (students[i], mark),
                 textcoords="offset points",
                 xytext=(0, 10),
                 ha='center',
                 fontsize=10,
                 fontweight='normal',
                 color='black')
plt.title("Scatter Plot: Student Marks")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.legend()
plt.show()
```
